```python
import numpy as np
import time

from robosuite import make
from robosuite.controllers import controller_factory, load_part_controller_config

env = make(
    env_name="Lift",
    robots="Panda",
    # controller_configs=controller,
    has_renderer=True
)


obs = env.reset()
print(obs.keys())  # See available keys

cube_pos = obs.get("cube_pos")  # Adjust key based on your env
print(f"Cube Position: {cube_pos}")
```

```python
# Example: Randomize object color
num_geoms = env.sim.model.geom_rgba.shape[0]
print(num_geoms)
for geom_id in range(num_geoms):
    geom_name = env.sim.model.geom_id2name(geom_id)
    if "cube" in geom_name or "object" in geom_name:  # adjust based on your task's geom names
        print(f'The cube geometry \n name: {geom_name} \n id: {geom_id}')

'''
The cube geometry 
 name: cube_g0 
 id: 88
The cube geometry 
 name: cube_g0_vis 
 id: 89
'''
```

```python
# Example: Random RGBA color
random_color = np.concatenate([np.random.rand(3), [1.0]])  # RGB random, alpha=1

cube_geom_ids = [
    env.sim.model.geom_name2id("cube_g0"),
    env.sim.model.geom_name2id("cube_g0_vis")
]

for geom_id in cube_geom_ids:
    env.sim.model.geom_rgba[geom_id] = random_color

env.sim.forward()

geom_id = 88
print(env.sim.model.geom_rgba[geom_id])
```

```python
cube_body_id = env.sim.model.body_name2id("cube_main")

print(cube_body_id)

print("Cube Position:", env.sim.model.body_pos[cube_body_id])
print("Cube Orientation (quat):", env.sim.model.body_quat[cube_body_id])
```
